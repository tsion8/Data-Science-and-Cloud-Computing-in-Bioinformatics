```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_FOLDER = r"C:\Users\MY-LENOVO\OneDrive\Desktop\processed"

DATA_FILE = os.path.join(
    DATA_FOLDER,
    "GSE52778_All_Sample_FPKM_Matrix.txt.gz"
)

expr = pd.read_csv(
    DATA_FILE,
    sep=r"\s+",
    compression="gzip"
)

print("Expression data loaded successfully!")
print("Dataset shape:", expr.shape)
print("\nColumn names:")
print(expr.columns.tolist())
print("\nFirst five rows:")
display(expr.head())
```

```python
expression_samples = [
    "Dex_LL14",
    "Dex_LL06",
    "Dex_LL02",
    "Dex_LL10",
    "Untreated_LL09",
    "Untreated_LL05",
    "Untreated_LL13",
    "Untreated_LL01"
]

required_columns = ["gene_id", "gene_short_name"] + expression_samples

missing_columns = [
    col for col in required_columns if col not in expr.columns
]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

analysis_expr = expr[required_columns].copy()

print("Selected expression data:", analysis_expr.shape)
print("Number of selected samples:", len(expression_samples))
display(analysis_expr.head())
```

```python
metadata = pd.DataFrame({
    "Sample": expression_samples,
    "Treatment": [
        "DEX", "DEX", "DEX", "DEX",
        "Untreated", "Untreated", "Untreated", "Untreated"
    ],
    "Donor": [
        "N61311", "N052611", "N080611", "N061011",
        "N61311", "N052611", "N080611", "N061011"
    ]
})

metadata["Treatment"] = pd.Categorical(
    metadata["Treatment"],
    categories=["Untreated", "DEX"]
)

print("Sample metadata:")
display(metadata)
```

```python
expression_ids = expression_samples
metadata_ids = metadata["Sample"].tolist()

print("Expression sample count:", len(expression_ids))
print("Metadata sample count:", len(metadata_ids))
print("Sample identifiers match:", set(expression_ids) == set(metadata_ids))
print("Duplicate metadata identifiers:", metadata["Sample"].duplicated().sum())

print("Expression samples missing from metadata:",
      sorted(set(expression_ids) - set(metadata_ids)))

print("Metadata samples missing from expression:",
      sorted(set(metadata_ids) - set(expression_ids)))
```

```python
print("Missing values per selected column:")
display(analysis_expr.isna().sum().to_frame("Missing_Count"))

print("Total missing values:", analysis_expr.isna().sum().sum())
print("Duplicate rows:", analysis_expr.duplicated().sum())
print("Duplicate gene IDs:", analysis_expr["gene_id"].duplicated().sum())

# Check whether expression values are numeric and valid
expression_values = analysis_expr[expression_samples].apply(
    pd.to_numeric, errors="coerce"
)

print("Non-numeric expression values:", expression_values.isna().sum().sum())
print("Negative expression values:", (expression_values < 0).sum().sum())

if not np.isfinite(expression_values.to_numpy()).all():
    print("Warning: infinite or non-finite expression values detected.")
else:
    print("All expression values are finite.")

analysis_expr[expression_samples] = expression_values
```

```python
expression_matrix = analysis_expr.set_index(
    ["gene_id", "gene_short_name"]
)[expression_samples].copy()

print("Genes:", expression_matrix.shape[0])
print("Samples:", expression_matrix.shape[1])
display(expression_matrix.head())
```

```python
total_fpkm = expression_matrix.sum(axis=0)

print("Total FPKM per sample:")
display(total_fpkm.to_frame("Total_FPKM"))

print(
    "\nNote: true sequencing library size must be calculated "
    "from raw read counts."
)
```

```python
detected_genes = (expression_matrix > 0).sum(axis=0)

sample_qc = pd.DataFrame({
    "Sample": expression_samples,
    "Total_FPKM": total_fpkm.reindex(expression_samples).values,
    "Detected_Genes": detected_genes.reindex(expression_samples).values
})

sample_qc = sample_qc.merge(metadata, on="Sample", how="left")

display(sample_qc)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Transform expression values
log_expression = np.log2(expression_matrix + 1)

# Identify the 500 most variable genes
gene_variance = log_expression.var(axis=1)

hvg_table = pd.DataFrame({
    "gene_id": log_expression.index.get_level_values("gene_id"),
    "gene_short_name": log_expression.index.get_level_values("gene_short_name"),
    "variance": gene_variance.values
}).sort_values("variance", ascending=False)

top_hvg_ids = hvg_table.head(500)["gene_id"].tolist()

pca_expression = log_expression[
    log_expression.index.get_level_values("gene_id").isin(top_hvg_ids)
]

# Samples must be rows and genes must be columns
pca_input = pca_expression.T

# Standardize and perform PCA
scaler = StandardScaler()
pca_scaled = scaler.fit_transform(pca_input)

pca = PCA(n_components=2)
pca_coordinates = pca.fit_transform(pca_scaled)

# Create PCA results table
pca_df = pd.DataFrame(
    pca_coordinates,
    columns=["PC1", "PC2"],
    index=pca_input.index
)

pca_df.index.name = "Sample"
pca_df = pca_df.reset_index()

# Attach treatment and donor information
pca_df = pca_df.merge(
    metadata[["Sample", "Treatment", "Donor"]],
    on="Sample",
    how="left"
)

pc1_variance = pca.explained_variance_ratio_[0] * 100
pc2_variance = pca.explained_variance_ratio_[1] * 100

print("PCA results recreated successfully.")
print(f"PC1 explains {pc1_variance:.2f}% of variance.")
print(f"PC2 explains {pc2_variance:.2f}% of variance.")

display(pca_df)
```

```python
plt.figure(figsize=(8, 6))

for donor in pca_df["Donor"].unique():
    subset = pca_df[pca_df["Donor"] == donor]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        s=90,
        label=donor
    )

    for _, row in subset.iterrows():
        plt.annotate(
            row["Sample"],
            (row["PC1"], row["PC2"]),
            fontsize=8,
            xytext=(4, 4),
            textcoords="offset points"
        )

plt.xlabel(f"PC1 ({pc1_variance:.1f}% variance)")
plt.ylabel(f"PC2 ({pc2_variance:.1f}% variance)")
plt.title("PCA Colored by Donor/Cell Line")
plt.legend(title="Donor", fontsize=8)
plt.tight_layout()

plt.savefig(
    os.path.join(FIGURES_DIR, "06_PCA_by_donor.png"),
    dpi=300
)
plt.show()
```

```python
pca_df["PCA_Distance"] = np.sqrt(
    pca_df["PC1"]**2 + pca_df["PC2"]**2
)

outlier_table = pca_df[
    ["Sample", "Treatment", "Donor", "PC1", "PC2", "PCA_Distance"]
].sort_values(
    "PCA_Distance",
    ascending=False
)

print("Samples ranked by PCA distance:")
display(outlier_table)
```
